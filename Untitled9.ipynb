```python
import pandas as pd

df = pd.read_csv("captions.txt")

print(df.head())
print("Total captions:", len(df))
print("Unique images:", df["image"].nunique())

mapping = {}

for _, row in df.iterrows():
    image_id = row["image"]
    caption = row["caption"]

    if image_id not in mapping:
        mapping[image_id] = []

    mapping[image_id].append(caption)

print("Total mapped images:", len(mapping))
```

```python
def clean_caption(caption):
    caption = caption.lower()
    caption = ''.join(char for char in caption if char.isalpha() or char.isspace())
    caption = ' '.join(caption.split())
    caption = "startseq " + caption + " endseq"
    return caption

for image_id, captions in mapping.items():
    mapping[image_id] = [clean_caption(caption) for caption in captions]

first_image = list(mapping.keys())[0]
print(first_image)
print(mapping[first_image])
```

```python
from tensorflow.keras.preprocessing.text import Tokenizer
import pickle

all_captions = []

for captions in mapping.values():
    for caption in captions:
        all_captions.append(caption)

tokenizer = Tokenizer()
tokenizer.fit_on_texts(all_captions)

vocab_size = len(tokenizer.word_index) + 1
max_length = max(len(caption.split()) for caption in all_captions)

print("Vocabulary size:", vocab_size)
print("Max caption length:", max_length)

with open("tokenizer.pkl", "wb") as f:
    pickle.dump(tokenizer, f)
```

```python
import numpy as np
import pickle
import os
from tqdm import tqdm

from tensorflow.keras.applications.xception import Xception, preprocess_input
from tensorflow.keras.preprocessing.image import load_img, img_to_array
from tensorflow.keras.models import Model

base_model = Xception(weights="imagenet")
feature_extractor = Model(
    inputs=base_model.input,
    outputs=base_model.layers[-2].output
)

features = {}

image_names = list(mapping.keys())

for img_name in tqdm(image_names):
    img_path = os.path.join(image_folder, img_name)

    if not os.path.exists(img_path):
        continue

    image = load_img(img_path, target_size=(299, 299))
    image = img_to_array(image)
    image = np.expand_dims(image, axis=0)
    image = preprocess_input(image)

    feature = feature_extractor.predict(image, verbose=0)
    features[img_name] = feature[0]

print("Extracted features for:", len(features), "images")

with open("features.pkl", "wb") as f:
    pickle.dump(features, f)
```

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.utils import to_categorical
import numpy as np

def data_generator(mapping, features, tokenizer, max_length, vocab_size, batch_size):
    X1, X2, y = [], [], []
    n = 0

    while True:
        for image_id, captions in mapping.items():
            if image_id not in features:
                continue

            image_feature = features[image_id]

            for caption in captions:
                seq = tokenizer.texts_to_sequences([caption])[0]

                for i in range(1, len(seq)):
                    in_seq = seq[:i]
                    out_seq = seq[i]

                    in_seq = pad_sequences([in_seq], maxlen=max_length)[0]
                    out_seq = to_categorical([out_seq], num_classes=vocab_size)[0]

                    X1.append(image_feature)
                    X2.append(in_seq)
                    y.append(out_seq)

                    n += 1

                    if n == batch_size:
                        X1_arr = np.array(X1, dtype=np.float32)
                        X2_arr = np.array(X2, dtype=np.int32)
                        y_arr = np.array(y, dtype=np.float32)

                        # IMPORTANT: tuple, not list
                        yield ((X1_arr, X2_arr), y_arr)

                        X1, X2, y = [], [], []
                        n = 0
```

```python
from tensorflow.keras.layers import Input, Dense, LSTM, Embedding, Dropout, add
from tensorflow.keras.models import Model

# Image feature input
inputs1 = Input(shape=(2048,))
fe1 = Dropout(0.5)(inputs1)
fe2 = Dense(256, activation="relu")(fe1)

# Text sequence input
inputs2 = Input(shape=(max_length,))
se1 = Embedding(vocab_size, 256, mask_zero=True)(inputs2)
se2 = Dropout(0.5)(se1)
se3 = LSTM(256)(se2)

# Combine image + text
decoder1 = add([fe2, se3])
decoder2 = Dense(256, activation="relu")(decoder1)
outputs = Dense(vocab_size, activation="softmax")(decoder2)

caption_model = Model(inputs=[inputs1, inputs2], outputs=outputs)

caption_model.compile(
    loss="categorical_crossentropy",
    optimizer="adam"
)

caption_model.summary()
```

```python
# Use a smaller subset so training does not take forever
subset_size = 500   # try 300 first if you want even faster

small_mapping = {}
count = 0

for image_id, captions in mapping.items():
    if image_id in features:
        small_mapping[image_id] = captions
        count += 1

    if count == subset_size:
        break

print("Using images:", len(small_mapping))


batch_size = 64
epochs = 3   # use 2 or 3 first, not 20

# Count total training sequences for small dataset
total_sequences = 0

for image_id, captions in small_mapping.items():
    if image_id not in features:
        continue

    for caption in captions:
        seq = tokenizer.texts_to_sequences([caption])[0]
        total_sequences += len(seq) - 1

steps = total_sequences // batch_size

print("Total training sequences:", total_sequences)
print("Steps per epoch:", steps)

generator = data_generator(
    small_mapping,
    features,
    tokenizer,
    max_length,
    vocab_size,
    batch_size
)

history = caption_model.fit(
    generator,
    epochs=epochs,
    steps_per_epoch=steps,
    verbose=1
)
```

```python
import random
from PIL import Image
import matplotlib.pyplot as plt
import os
import numpy as np
from tensorflow.keras.preprocessing.sequence import pad_sequences

def word_for_id(integer, tokenizer):
    for word, index in tokenizer.word_index.items():
        if index == integer:
            return word
    return None

def generate_caption(model, tokenizer, image_feature, max_length):
    in_text = "startseq"

    for _ in range(max_length):
        sequence = tokenizer.texts_to_sequences([in_text])[0]
        sequence = pad_sequences([sequence], maxlen=max_length)

        yhat = model.predict(
            [image_feature.reshape(1, 2048), sequence],
            verbose=0
        )

        yhat = np.argmax(yhat)
        word = word_for_id(yhat, tokenizer)

        if word is None:
            break

        in_text += " " + word

        if word == "endseq":
            break

    final_caption = in_text.replace("startseq", "").replace("endseq", "").strip()
    return final_caption
```
